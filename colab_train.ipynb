```python
from ultralytics import YOLO
import os

last_weight = "runs/yolov9_smf_fpidet/weights/last.pt"

if os.path.exists(last_weight):
    model = YOLO(last_weight)
    model.train(resume=True)
else:
    model = YOLO("yolov9s.pt")
    model.train(
        data="data_smf_fpidet.yaml",
        epochs=100,
        imgsz=640,
        batch=-1,
        workers=8,
        device=0,
        project="runs",
        name="yolov9_smf_fpidet",
        save=True,
        save_period=1,
        val=True,
        cache=True,
        amp=True,
        optimizer="SGD",
        lr0=0.01,
        lrf=0.01,
        weight_decay=0.0005,
        warmup_epochs=3.0,
        seed=42,
        deterministic=True
    )
```
